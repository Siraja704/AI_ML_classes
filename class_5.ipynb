```python
from typing import List, Tuple
import statistics
```

```python
def fit_simple_linear_regression(x: List[float], y: List[float]) -> Tuple[float, float]:
   # y = m*x + b
 if len(x) != len(y):
          raise ValueError("Input lists must have the same length.")
 n = len(x)
 if n < 2:
         raise ValueError("At least two data points are required to fit a linear regression.")
    
 mean_x  = statistics.mean(x)
 mean_y  = statistics.mean(y)

 numinatotor = sum((xi - mean_x)* (yi - mean_y) for xi, yi in zip(x,y))
 demnominator = sum((xi -mean_x) **2 for xi in x)

 if demnominator == 0:
    raise ValueError("Variance of x is zero. Cannot fit a linear regression.")


 m = numinatotor / demnominator
 b = mean_y - m * mean_x
 return m , b 



     
```

```python
def predict(x: List[float], m: float, b: float) -> List[float]:
    return [m * xi + b for xi in x]
```

```python

x = [2,3,5,7,9,11,13,15,17,20]
y = [65,70,80,85,95,100,105,110,115,120]
```

```python
x = [2,3,5,7,9,11,13,15,17,20]
y = [65,70,80,85,95,100,105,110,115,120]

m, b = fit_simple_linear_regression(x, y)

x_test= [6, 8, 10]

predictions = predict(x_test, m, b)
print(f"Fitted line: y = {m:.2f}x + {b:.2f}")
print("Predictions:", predictions)
```
